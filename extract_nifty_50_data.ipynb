```python
import pandas as pd
import yfinance as yf
```

```python
nifty = yf.download("^NSEI", start="2022-01-01")['Close']

nifty_weekly = nifty.resample('W-FRI').last().dropna()
nifty_weekly
```

```python
nifty_weekly.columns
```

```python
nifty_df = nifty_weekly.reset_index()
nifty_df.columns = ["Date", "Nifty_50"]
nifty_df
```

```python
nifty_df.to_csv('data/nifty_50.csv', index=False)
```
