```python
for fund in mid_cap+small_cap+debt+gold+silver:
    if fund not in mf_df['Scheme NAV Name'].values:
        print(fund)
```

```python
fund_types = [mid_cap, small_cap, debt, gold, silver]
scheme_codes = {}
for fund_type in fund_types:
    for fund in fund_type:
        scheme_codes[fund] = mf_df[mf_df['Scheme NAV Name'] == fund]['Code'].values[0]
```

```python
with open('data/fund_names.json', 'w') as f:
    json.dump({
        "mid_cap": {int(scheme_codes[fund]): fund for fund in mid_cap},
        "small_cap": {int(scheme_codes[fund]): fund for fund in small_cap},
        "debt": {int(scheme_codes[fund]): fund for fund in debt},
        "gold": {int(scheme_codes[fund]): fund for fund in gold},
        "silver": {int(scheme_codes[fund]): fund for fund in silver}
    }, f)
```

```python
hist_nav_test = fetch_nav_history(scheme_codes["ICICI Prudential Silver ETF FOF - Direct Plan - Growth"], years=4, frequency='W')
hist_nav_test
```

```python
fund_df = build_portfolio_matrix(list(scheme_codes.values()), years=5, frequency='W')
```

```python
fund_df
```

```python
fund_df.isna().sum()
```

```python
fund_counts = {"mid_cap": 0, "small_cap": 0, "debt": 0, "gold": 0, "silver": 0}
for code, na_count in fund_df.isna().sum().items():
    if na_count >20:
        fund_name = mf_df[mf_df['Code'] == code]['Scheme NAV Name'].values[0]
        if fund_name in mid_cap:
            fund_counts["mid_cap"] += 1
        elif fund_name in small_cap:
            fund_counts["small_cap"] += 1
        elif fund_name in debt:
            fund_counts["debt"] += 1
        elif fund_name in gold:
            fund_counts["gold"] += 1
        elif fund_name in silver:
            fund_counts["silver"] += 1
        fund_df = fund_df.drop(columns=[code])
        print(f"Dropped {fund_name} due to {na_count} missing NAV data.")

print(fund_counts)
    
```
